```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt

#读数据的同时，直接把表头写进去(不要用usedcars的数据，太规范，不需要清洗，而且列数不对)
file_name=r"D:\pythonCourse\pythonCode\7204\M02\auto.csv"
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
df = pd.read_csv(file_name, names = headers)
df.head()

#从此开始，每次df执行操作，都要通过inplace=ture来进行覆盖原表
#数据清洗
df.replace("?", np.nan, inplace = True)
df.head(5)

#isnull处理 值只有True/False — 原来位置是 NaN 的变成 True，有值的变成 False。（所以missing_data是一个只包含false和true的矩阵）
missing_data = df.isnull()
missing_data.head(5)
print(missing_data)

#.columns → 获取所有列名（Index 对象）。.values → 将 Index 转为 NumPy 数组。.tolist() → 将 NumPy 数组转为 Python 列表。
#最终效果：遍历 missing_data 的每一个列名，依次赋值给变量 column。其实就是遍历列名
#三步转换，第一步，Index(['symboling', 'normalized-losses', 'make', 'fuel-type', ...], dtype='object')
#第二步，array(['symboling', 'normalized-losses', 'make', 'fuel-type', ...], dtype=object)
#第三步，['symboling', 'normalized-losses', 'make', 'fuel-type', ...]
for column in missing_data.columns.values.tolist():
    print(column)
    #用[]取列名对应的所有数值
    print (missing_data[column].value_counts()) #每个列名下面的true和false进行统计，你可以把 .value_counts() 理解成一个投票计数器，每个列名下面统计一下
    #换行
    print("")

#这两行代码的作用是计算 normalized-losses 列的平均值，为后续用均值填补缺失值做准备。
#df取normalized-losses列名下面的所有数的值
avg_norm_loss = df["normalized-losses"].astype("float").mean(axis=0) #.astype("float")，将该列转换为 float 类型。数字字符串 "122" → 浮点数 122.0；NaN 保持不变（NaN 本身就是 float 类型）；如果不转换，object 类型无法做数学运算
#在 Pandas 中，object 类型是一种"混合类型"的兜底数据类型。当一列中的数据无法用单一数值类型表示时，Pandas 就会将该列标记为 object。
# 122       ← 看起来是数字，但 Pandas 读进来是字符串 "122"
# 122
# ?         ← 这个 "?" 让它无法整体识别为数字
# 137
# 137

# .mean(axis=0) 计算该列的均值。axis=0 表示沿行方向（即对列求均值）。mean() 会自动跳过 NaN，只用有效值来计算。
# "122"  ─┐
# "122"   │  astype("float")   .mean()
# "?" → NaN ─┤  ──────────→  ──────────→ 122.0
# "137"   │    转为浮点数        跳过NaN求均值
# "137"  ─┘
print("Average of normalized-losses:", avg_norm_loss)

#replace 用每列的平均值来替换Nan
#这样写会触发警告机制,没有覆盖原df文件，不需要inplace=True
#df["normalized-losses"].replace(np.nan, avg_norm_loss, inplace=True)
df["normalized-losses"] = df["normalized-losses"].replace(np.nan, avg_norm_loss)


avg_bore=df['bore'].astype('float').mean(axis=0)
print("Average of bore:", avg_bore)
#df["bore"].replace(np.nan, avg_bore, inplace=True)
df["bore"]=df["bore"].replace(np.nan, avg_bore)
```
