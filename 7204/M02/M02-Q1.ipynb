```python
#第一步，用均值填充 stroke 列的 NaN（这是需要填写的内容）
avg_stroke = df['stroke'].astype('float').mean(axis=0)
print('Average of stroke:', avg_stroke)
df['stroke'] = df['stroke'].replace(np.nan, avg_stroke)

#继续填充
avg_horsepower = df['horsepower'].astype('float').mean(axis=0)
print("Average horsepower:", avg_horsepower)
df['horsepower'].replace(np.nan, avg_horsepower, inplace=True)
avg_peakrpm=df['peak-rpm'].astype('float').mean(axis=0)
print("Average peak rpm:", avg_peakrpm)
df['peak-rpm'].replace(np.nan, avg_peakrpm, inplace=True)

#统计一下车门的数量各有多少（车门这一列有两门的、有三门的）
df['num-of-doors'].value_counts()
#简单来说，.idxmax() 就是告诉你："谁出现得最多？" → 答案是 "four"，也就是四门车最多
#max()表示出现最多的个数
df['num-of-doors'].value_counts().idxmax()
#把车门的缺失值
df["num-of-doors"] = df["num-of-doors"].replace(np.nan, "four")
#作用：删除 price 列中有缺失值的整行数据。drop就是删除
df.dropna(subset=["price"], axis=0, inplace=True)
#作用：删除行之后，重新排列索引（0, 1, 2, 3, ...）。
#drop=True 表示删除旧索引，不保存成新的一列，如果drop=False,会保留旧索引，变成普通的一列
df.reset_index(drop=True, inplace=True)
df.head()
#显示每一列的数据类型
df.dtypes

#到这里还是数据清洗的一部分，就是将这些列下面的类型都转化为自己定义的类型
#这里注意，这种多个方括号
#单列操作用 df["col"]（Series）— 大多数情况；例如取均值 df["bore"].astype("float").mean()，例如删除行 → 基于某一列判断 df.dropna(subset=["price"], ...) 
#多列同时操作用 df[["col"]]，例如下面的同时转换多列的数据类型，这个返回的其实就是DF类型
df[["bore", "stroke"]] = df[["bore", "stroke"]].astype("float")
df[["normalized-losses"]] = df[["normalized-losses"]].astype("int")
df[["price"]] = df[["price"]].astype("float")
df[["peak-rpm"]] = df[["peak-rpm"]].astype("float")
df.dtypes

#这行代码的作用是将城市油耗从 mpg（英里/加仑）转换为 L/100km（升/百公里）。转换公式L/100km=235/mpg
#df['city-L/100km'] = 表示把计算结果作为新的一列添加到DF表中
#不覆盖原列，而是新建了一列 city-L/100km，原来的 city-mpg 保持不变
#属于数据清理里面的数据标准化内容
df.head()
df['city-L/100km'] = 235/df["city-mpg"]
df.head()




```
